# HR Attrition Analysis Insights

---

## 1. The "Income & Experience" Gap

### Inferences:

- Monthly Income:  
  Attrition is heavily concentrated among low earners (below 5,000/month). Once-salary-crosses 10,000/month, employees are significantly less likely to leave.

- Total Working Years:  
  Employees with 0–5 years of experience show the highest attrition rate, indicating loss of early-career talent before they become fully skilled.

- Years at Company:  
  A sharp spike in attrition is observed within the first year, suggesting issues with onboarding or expectation mismatch rather than long-term dissatisfaction.

---

### Suggestions:

- Entry-Level Salary Review:  
  Conduct a market benchmarking analysis for entry-level roles to ensure competitive compensation.

- Strong Onboarding (First 90 Days):  
  Improve onboarding experience to align job expectations with reality and strengthen cultural integration.

---

## 2. Demographic & Social Drivers

### Inferences:

- Marital Status:  
  Single employees have a significantly higher attrition rate compared to married or divorced employees.

- Distance from Home:  
  Employees living more than 10–12 units away are more likely to leave, likely due to commute fatigue.

- Gender:  
  While male employees dominate the workforce, attrition proportion is relatively similar across genders, with a slightly higher tendency among males.

- AgeGroup:
  18-25 and 26-35 age group have highly attrition rate. 

---

### Suggestions:

- Commute Support:  
  Introduce hybrid work options or transport facilities for employees with long travel distances.

- Employee Engagement:  
  Build stronger workplace communities such as clubs, events, and team activities to improve retention among single & young employees.

---

## 3. Workplace & Satisfaction Drivers

### Inferences:

- Department:  
  Sales and R&D show notable attrition levels. Sales roles often experience higher turnover due to pressure and target-driven environments.

- Job Satisfaction:  
  Employees with satisfaction levels 1 (Low) and 2 (Medium) show the highest attrition. However, even some satisfied employees leave, indicating that career growth also plays a key role.

---

### Suggestions:

- Sales Burnout Analysis:  
  Review workload, incentives, and performance pressure in the Sales department.

- Stay Interviews:  
  Conduct proactive Stay Interviews for employees with low satisfaction to address issues before resignation occurs.

---

## Summary (Dashboard Insight)

### Key Risk Profile:
A single employee, earning below $5,000, with less than 2 years of experience, and a long commute distance.

---

### Top Recommendation:
Focus on improving compensation for low-income employees and offering flexible or hybrid work options for long-distance commuters. These two interventions can significantly reduce overall attrition.

# SQL For Analysis, Data Cleaning & KPI Generation

```python
import sqlite3

conn = sqlite3.connect('df.db') #database created
```

```python
df.to_sql('hr', conn, index = False, if_exists='replace') #hr name table created
```

```python
pd.read_sql("""Select * from hr LIMIT 5""",conn)
```

### Queries for data insights

```python
#Top 5 highest MonthlyIncome employee details.

pd.read_sql("""Select EmpID, AgeGroup, EducationField, Department, JobRole, MonthlyIncome from hr
Order by MonthlyIncome Desc
Limit 5""",conn)
```

```python
pd.read_sql("""Select Attrition, count(Attrition) AS Total_Attrition 
FROM hr
GROUP BY Attrition""",conn)
```

#### Employee Attrition by JobRole

```python
pd.read_sql("""
SELECT JobRole,
       COUNT(*) AS Total_Att
FROM hr
WHERE Attrition = 'Yes'
GROUP BY JobRole
ORDER BY Total_Att DESC
""", conn)
```

#### Employee Attrition by Department

```python
pd.read_sql("""
SELECT Department,
       COUNT(*) AS Total_Att
FROM hr
WHERE Attrition = 'Yes'
GROUP BY Department
ORDER BY Total_Att DESC
""", conn)
```

#### Top 10 Employees with Highest Years at Company

```python
pd.read_sql("""SELECT EmpID, YearsAtCompany
FROM hr
ORDER BY YearsAtCompany DESC
LIMIT 10""",conn)
```

#### Top 10 Employees with Highest Years in Current Role

```python
pd.read_sql("""SELECT EmpID, YearsInCurrentRole
from hr
Order By YearsInCurrentRole  DESC
Limit 10""",conn)
```

#### Top 10 Employees by Years Since Last Promotion.

```python
pd.read_sql("""SELECT EmpID, YearsSinceLastPromotion
from hr
Order By YearsSinceLastPromotion  DESC
Limit 10""",conn)
```

#### Top 10 Employees by Years with Current Manager

```python
pd.read_sql("""SELECT EmpID, YearsWithCurrManager
from hr
Order By YearsWithCurrManager  DESC
Limit 10""",conn)
```

#### Department-wise Count of Employees Who Travel Rarely

```python
pd.read_sql("""
SELECT Department,
       COUNT(*) AS Total_Travel_Rarely
FROM hr
WHERE BusinessTravel = 'Travel_Rarely'
GROUP BY Department
ORDER BY Total_Travel_Rarely DESC
""", conn)
```

#### Department-wise Count of Employees Who Travel_Frequently

```python
pd.read_sql("""
SELECT Department,
       COUNT(*) AS Total_Travel_Frequently
FROM hr
WHERE BusinessTravel = 'Travel_Frequently'
GROUP BY Department
ORDER BY Total_Travel_Frequently DESC
""", conn)
```

#### Department-wise Count of Employees Who Not Travel.

```python
pd.read_sql("""
SELECT Department,
       COUNT(*) AS Total_Non_Travel	
FROM hr
WHERE BusinessTravel = 'Non-Travel'
GROUP BY Department
ORDER BY Total_Non_Travel DESC
""", conn)
```

#### Gender-wise Employee Count

```python
pd.read_sql("""
SELECT Gender,
       COUNT(*) AS Total
FROM hr
GROUP BY Gender
""", conn)
```

#### Gender-wise Employee Count by Department

```python
pd.read_sql("""SELECT Department,Gender,Count(*) AS Total
FROM hr
GROUP BY Department, Gender
ORDER BY Department""",conn)
```

#### Employee Count by Job Role

```python
pd.read_sql("""SELECT JobRole,Count(*) AS Total
from hr
GROUP BY JobRole
ORDER BY Total DESC""", conn)
```
